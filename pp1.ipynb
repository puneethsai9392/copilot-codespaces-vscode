```python
#16
n=10
n1=0
n2=1
print(f'Fibonacci series: {n1} {n2}',end=' ')
for i in range(2,n+1):
    n3=n1+n2
    n1=n2
    n2=n3
    print(n3,end=' ')
print()
```

```python
#recursion
def fibonacciseries(i):
    if i<=1:
        return i
    else:
        return (fibonacciseries(i-1)+fibonacciseries(i-2))
num=10
if num<=0:
    print('Enter a valid number')
else:
    print('Fibonacciseries:',end=' ')
    for i in range(num):
        print(fibonacciseries(i),end=' ')
```
